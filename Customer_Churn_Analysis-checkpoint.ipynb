```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("WA_Fn-UseC_-Telco-Customer-Churn.csv")
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
```

```python
df.dropna(inplace=True)
```

```python
df.info()
```

```python
df.shape[0]
```

```python
df['Churn'].value_counts()
```

```python
(df['Churn'].value_counts(normalize=True))*100
```

```python
contract_churn = pd.crosstab(df['Contract'], df['Churn'])
contract_churn
```

```python
df.groupby("Churn")["MonthlyCharges"].mean()
```

```python
payment_churn = pd.crosstab(df['PaymentMethod'], df['Churn'])
payment_churn
```

```python
senior_churn = pd.crosstab(df['SeniorCitizen'], df['Churn'])
senior_churn
```

```python
sns.countplot(x='Churn', data=df)
plt.title("Customer Churn Count")
plt.savefig("Customer Churn Count.png")
plt.show()
```

```python
sns.countplot(x='Contract', hue='Churn', data=df)
plt.title("Contract Type vs Churn")
plt.savefig("Contract Type vs Churn.png")
plt.show()
```

```python
sns.boxplot(x='Churn', y='MonthlyCharges', data=df)
plt.title("Monthly Charges vs Churn")
plt.savefig("Monthly Charges vs Churn.png")
plt.show()
```

```python
sns.countplot(x='PaymentMethod', hue='Churn', data=df)
plt.xticks(rotation=45)
plt.title("Payment Method vs Churn")
plt.savefig("Payment Method vs Churn.png")
plt.show()
```

```python
sns.countplot(x='InternetService', hue='Churn', data=df)

plt.title("Internet Service vs Churn")
plt.xlabel("Internet Service")
plt.ylabel("Customer Count")
plt.savefig("Internet Service vs Churn.png")
plt.show()
```
